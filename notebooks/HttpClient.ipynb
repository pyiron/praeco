# HttpClient

`praeco.HttpClient` is a building block used by service clients whose API builds on top of http requests (e.g. `OntodockerClient`). It can also be used directly for services where no proper client is implemented yet.

This notebook shows the minimal usage of its public `request(...)` escape hatch, allowing raw requests within `praeco`. This can be used, e.g., in the case a specific service exposes endpoints which are not wrapped by a client implementation available inside `praeco`.

## Setup
To execute this notebook, your shell needs to export two environment variables:
```bash
export HTTPCLIENT_ADDRESS="<address of your http service's endpoint>"
export HTTPCLIENT_TOKEN="<bearer token for that service>"
```
The request below calls Ontodocker's `/api/v1/endpoints` route, so point `HTTPCLIENT_ADDRESS` to an Ontodocker instance to run it unchanged.

Then, launch Jupyter from this shell.

````python
import os


def require_env(name: str) -> str:
    value = os.getenv(name)
    if value is None or not value.strip():
        raise RuntimeError(f"Required environment variable {name} is not set.")
    return value.strip()
````

````python
token = require_env("HTTPCLIENT_TOKEN")
address = require_env("HTTPCLIENT_ADDRESS")
````

````python
from praeco import HttpClient

client = HttpClient(address=address, token=token)
client.base_url
````

````python
resp = client.request(method="GET", url=client.base_url + "/api/v1/endpoints")
resp.status_code
````
